```python
unique_item_map = (
    item_map[item_map.apply(len) == 1]
    .apply(lambda x: x[0])
    .to_dict()
)

unique_item_map
```

```python
mask = (
    (df['Item'] == 'UNKNOWN') &
    (df['Price Per Unit'].isin(unique_item_map.keys()))
)

df.loc[mask, 'Item'] = (
    df.loc[mask, 'Price Per Unit']
      .map(unique_item_map)
)
```

```python
# Check: Quantity × Price Per Unit = Total Spent
cols = ["Quantity", "Price Per Unit", "Total Spent"]
check = df[cols].apply(pd.to_numeric, errors="coerce")

valid = check.notna().all(axis=1)
calculated_total = check["Quantity"] * check["Price Per Unit"]

matches = np.isclose(
    calculated_total[valid],
    check.loc[valid, "Total Spent"],
    rtol=1e-5,
    atol=1e-8
)

print(f"Rows checked: {valid.sum()}")
print(f"Rows matching the formula: {matches.sum()}")
print(f"Rows not matching: {(~matches).sum()}")

# Display any rows that do not match
mismatch_rows = df.loc[valid].loc[~matches]
mismatch_rows
```

```python
price_to_item_count = (
    df[df["Item"].notna() & (df["Item"] != "UNKNOWN")]
    .groupby("Price Per Unit")["Item"]
    .nunique()
)

unique_prices = price_to_item_count[price_to_item_count == 1]
shared_prices = price_to_item_count[price_to_item_count > 1]

print("Prices that identify one item:")
print(unique_prices)

print("\nPrices shared by multiple items:")
print(shared_prices)
```
